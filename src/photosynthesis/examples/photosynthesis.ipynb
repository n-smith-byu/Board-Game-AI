### Quick Start

Runs a simple game with 3 random players. Outputs the final scores.

```python
from photosynthesis import PhotosynthesisGame, RandomPlayer

game = PhotosynthesisGame(players=[RandomPlayer(i) for i in range(3)], extra_round=False)
game.reset()
game.run()
```

### Using a Waiting Room

```python
from photosynthesis import PhotosynthesisGame, RandomPlayer
from board_game import WaitingRoom

PhotosynthesisGame.set_ai_player_class(RandomPlayer)        # Optional (RandomPlayer is the default)

wr = WaitingRoom(PhotosynthesisGame)

wr.add_bot()
wr.add_player('Player2')
wr.add_bot()

print(f'Num Humans: {wr.num_human_players()}')
print(f'Num Bots: {wr.num_bots()}')

game = wr.create_game()
```

```python
# Run to play the game through the terminal.
# (Not Recommended in this Jupyter Notebook)
game.run(display=True)          # display tells the game whether to print the board out. It should tpyically be used whenever a human player is involved.
```
